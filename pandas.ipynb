```python
# 创建示例数据
data = {
   'department': ['HR', 'Tech', 'HR', 'Tech'],
        'salary': [6000, 8000, 7000, 9000],
        'experience': [3, 5, 4, 6]}
df = pd.DataFrame(data)

# 定义自定义聚合函数
def custom_agg(series):
    return series.max() - series.min()

# 按部门分组并应用自定义聚合函数
custom_agg_result = df.groupby('department')['salary'].agg(custom_agg)
print("按部门分组并应用自定义聚合函数：")
print(custom_agg_result)
```

## 通过字典进行分组统计，把北京、上海、广州三个城市放在一起统计

```python
df = pd.read_csv("./data/JD01.csv",encoding='gbk')
df.set_index("商品名称",inplace=True)
dict1 = {"北京出库销量":"北上广","上海出库销量":"北上广","广州出库销量":"北上广","成都出库销量":"成都","武汉出库销量":"武汉"}
df1 = df.groupby(dict1,axis='columns').sum()
df1
```

## 计算和上一次的升降

```python
data = [110,105,125,137,116]
index = [1,2,3,4,5]

df3 = pd.DataFrame(data,index=index,columns=['语文'])
df3['和上次升降'] = df3 - df3.shift()
df3
```
